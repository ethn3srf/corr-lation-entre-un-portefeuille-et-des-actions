```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
tickers = [
    "BESI.AS",
    "ALBIO.PA",
    "ENGI.PA",
    "MAU.PA",
    "ALRIB.PA",
    "STMPA.PA",
    "HO.PA",
    "VLA.PA",
    "LQQ.PA",
    "EXA1.AS",
    "ETL.PA",
]
```

```python
weights = np.array([
    0.0664653,  # BESI
    0.0166811,  # Biosynex
    0.0857736,  # Engie
    0.0352324,  # Maurel & Prom
    0.0288497,  # Riber
    0.0681581,  # STMicro
    0.0887683,  # Thales
    0.0151301,  # Valneva
    0.228948,  # LQQ
    0.339815,  # EXA1
    0.0243977, # Eutelsat
])
```

```python
prices = yf.download(
    tickers,
    start="2021-12-24",
    end="2026-09-15",
    auto_adjust=True
)["Close"]
```

```python
prices = prices.drop(pd.Timestamp("2026-09-07"))
prices = prices.drop(pd.Timestamp("2025-10-24"))
```

```python
#print(prices.columns)
print(weights.sum())
print(prices.isna().sum())
```

```python
 prices[prices.isna().any(axis=1)] 
```

```python
returns = prices.pct_change().dropna()
cov_matrix = returns.cov()
cov_matrix

portfolio_variance = weights.T @ cov_matrix @ weights
portfolio_variance

portfolio_volatility = np.sqrt(portfolio_variance) 
portfolio_volatility
```

```python
annual_portfolio_volatility = portfolio_volatility * np.sqrt(252)
print(f"{annual_portfolio_volatility * 100} %")
```
